```python
import pandas as pd
import numpy as np

from pathlib import Path
```

```python
DATA_PATH = Path("../data/raw/application_train.csv")
df = pd.read_csv(DATA_PATH)
```

```python
dtype_summary = (
    df.dtypes
      .astype(str)
      .value_counts()
      .reset_index()
)

dtype_summary.columns = ["Data Type", "Count"]

dtype_summary
```

```python
numerical_features = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_features = df.select_dtypes(include=["object"]).columns.tolist()
boolean_features = df.select_dtypes(include=["bool"]).columns.tolist()

print(f"Numerical Features : {len(numerical_features)}")
print(f"Categorical Features : {len(categorical_features)}")
print(f"Boolean Features : {len(boolean_features)}")
```

```python
# ==========================================
# Section 6 : Memory Usage
# ==========================================

memory_usage = (
    df.memory_usage(deep=True)
      .sort_values(ascending=False)
      .reset_index()
)

memory_usage.columns = ["Column", "Memory (Bytes)"]

memory_usage["Memory (MB)"] = memory_usage["Memory (Bytes)"] / (1024**2)

memory_usage.head(10)
```

```python
# ==========================================
# Section 7 : Duplicate Analysis
# ==========================================

duplicate_rows = df.duplicated().sum()

print(f"Duplicate Rows : {duplicate_rows}")
```

```python
duplicate_ids = df["SK_ID_CURR"].duplicated().sum()

print(f"Duplicate Customer IDs : {duplicate_ids}")
```

```python
# ==========================================
# Section 8 : Target Variable Analysis
# ==========================================
target_summary = (
    df["TARGET"]
    .value_counts()
    .rename_axis("TARGET")
    .reset_index(name="Count")
)

target_summary["Percentage"] = (
    target_summary["Count"] / len(df)
) * 100

target_summary
```

```python
majority = target_summary["Count"].max()
minority = target_summary["Count"].min()

print(f"Imbalance Ratio : {majority/minority:.2f} : 1")
```

```python
# ==========================================
# Section 9 : Missing Value Analysis
# ==========================================

missing_summary = pd.DataFrame({

    "Feature": df.columns,
    "Missing Count": df.isnull().sum().values,

    "Missing Percentage":
    (df.isnull().sum().values / len(df)) * 100

})

missing_summary = missing_summary.sort_values(
    by="Missing Percentage",
    ascending=False
)

missing_summary.head(20)
```

```python
missing_summary[
    missing_summary["Missing Percentage"] > 50
]
```

```python
feature_categories = {

    "Target": ["TARGET"],

    "Identifiers": [
        "SK_ID_CURR"
    ],

    "Loan Features": [
        col for col in df.columns
        if "AMT_" in col
    ],

    "External Scores": [
        col for col in df.columns
        if "EXT_SOURCE" in col
    ],

    "Document Flags": [
        col for col in df.columns
        if "FLAG_DOCUMENT" in col
    ],

    "Region Features": [
        col for col in df.columns
        if "REGION" in col
    ],

    "Time Features": [
        col for col in df.columns
        if "DAYS" in col
    ]
}
# print(feature_categories)
for category, cols in feature_categories.items():
    print(f"{category} ({len(cols)})")
    print(cols)
    print("-"*60)
```

```python
# ==========================================
# Numerical Feature Analysis
# ==========================================

int_features = df.select_dtypes(include="int64").columns.tolist()

float_features = df.select_dtypes(include="float64").columns.tolist()

print(f"Integer Features : {len(int_features)}")

print(f"Float Features : {len(float_features)}")
```

```python
time_features = [
    col for col in df.columns
    if "DAYS" in col
]

time_features
```

```python
cardinality = pd.DataFrame({

    "Feature": categorical_features,

    "Unique Values":
    [df[col].nunique() for col in categorical_features]

})

cardinality.sort_values(
    by="Unique Values",
    ascending=False
)
```

```python
binary_features = [
    col
    for col in categorical_features
    if df[col].nunique()==2
]

binary_features
```

```python
identifier_columns = [

    col
    for col in df.columns
    if "SK_ID" in col
]

identifier_columns
```

```python
high_cardinality = [

    col

    for col in categorical_features

    if df[col].nunique() > 20

]

high_cardinality
```

```python
quality_report = pd.DataFrame({

    "Check":[

        "Missing Values",

        "Duplicate Rows",

        "Duplicate IDs",

        "Target Available",

        "Mixed Data Types",

        "High Cardinality"

    ],

    "Status":[

        "✔️" if df.isnull().sum().sum()>0 else "❌",

        "✔️" if duplicate_rows==0 else "❌",

        "✔️" if duplicate_ids==0 else "❌",

        "✔️" if "TARGET" in df.columns else "❌",

        "✔️",

        "✔️" if len(high_cardinality)>0 else "❌"

    ]

})

quality_report
```

## Key Observations

- The dataset contains approximately 307,000 loan applications with 122 features.
- The target variable (`TARGET`) is binary and exhibits class imbalance, so evaluation metrics such as ROC-AUC, Precision, Recall, F1-score, and PR-AUC will be more informative than accuracy.
- Several features contain a high percentage of missing values and will require careful handling during preprocessing.
- The dataset includes both numerical and categorical variables, requiring different preprocessing strategies.
- `SK_ID_CURR` is a unique customer identifier and should not be used directly as a predictive feature. It will later serve as the primary key for joining additional datasets.
- Features such as `EXT_SOURCE_1`, `EXT_SOURCE_2`, and `EXT_SOURCE_3` are expected to be highly predictive and deserve closer examination during EDA.
- Time-related variables (e.g., `DAYS_BIRTH`, `DAYS_EMPLOYED`) are stored as negative day offsets and will need transformation into meaningful units such as years.
- The presence of high-cardinality and near-constant features suggests that feature selection will be an important step before model training.
