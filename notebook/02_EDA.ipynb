- Large loans may represent higher financial risk.

### AMT_GOODS_PRICE

## Business Interpretation

Financial variables capture the applicant's repayment capacity.

Examples:

- Higher income generally increases repayment ability.
- Higher credit amounts increase financial exposure.
- High annuity relative to income may indicate repayment stress.
- Goods price provides context for the requested loan amount.

These variables are expected to play a major role in credit risk prediction.

### Observations

- All financial variables exhibit positive skewness, indicating that most applicants have moderate financial values while a small number have exceptionally large values.
- `AMT_INCOME_TOTAL` shows extremely high skewness and kurtosis, suggesting the presence of a few very large income values.
- These extreme observations should be investigated before applying any transformation or outlier treatment.

### Business Interpretation

The right-skewed distributions are consistent with real-world financial data, where a small number of applicants possess substantially higher income or request larger loans than the majority.

### Engineering Decisions

- Inspect the largest income values to determine whether they are valid observations or data quality issues.
- Consider log transformation for highly skewed financial variables.
- Postpone any outlier removal until dedicated outlier analysis, as extreme values may represent genuine high-net-worth customers.

Time feature

| Feature           | Meaning                                                       |
| ----------------- | ------------------------------------------------------------- |
| DAYS_BIRTH        | Days before application when the applicant was born           |
| DAYS_EMPLOYED     | Days before application when current employment started       |
| DAYS_REGISTRATION | Days since the applicant registered at the current residence  |
| DAYS_ID_PUBLISH   | Days since the applicant's identity document was last updated |


```python
time_features = [
    "DAYS_BIRTH",
    "DAYS_EMPLOYED",
    "DAYS_REGISTRATION",
    "DAYS_ID_PUBLISH"
]

df[time_features].head()
```

```python
summary = df[time_features].describe().T
summary["Missing"] = df[time_features].isnull().sum()

summary["Missing %"] = (
    summary["Missing"] / len(df) * 100
).round(2)

summary
```

```python
time_df = pd.DataFrame()

time_df["AGE_YEARS"] = (-df["DAYS_BIRTH"] / 365.25).round(1)

time_df["EMPLOYMENT_YEARS"] = (-df["DAYS_EMPLOYED"] / 365.25).round(1)

time_df["REGISTRATION_YEARS"] = (-df["DAYS_REGISTRATION"] / 365.25).round(1)

time_df["ID_PUBLISH_YEARS"] = (-df["DAYS_ID_PUBLISH"] / 365.25).round(1)

time_df.head()
```

```python
for column in time_df.columns:

    fig, axes = plt.subplots(1, 2, figsize=(12,4))

    time_df[column].plot(
        kind="hist",
        bins=50,
        ax=axes[0]
    )

    axes[0].set_title(f"{column} Distribution")

    time_df.boxplot(
        column=column,
        ax=axes[1]
    )

    axes[1].set_title(f"{column} Boxplot")

    plt.tight_layout()

    plt.show()
```
